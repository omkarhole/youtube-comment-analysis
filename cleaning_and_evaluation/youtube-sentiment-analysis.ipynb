```python
import html ,re

# Text preprocessing
def preprocess_text(text):
    text = str(text)

    # Decode HTML entities
    text = html.unescape(text)

    # Remove HTML tags where present
    text = re.sub(r"<[^>]+>", " ", text)

    # Replace URLs with a token
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " URL ",
        text
    )

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
stop_words = set(stopwords.words("english"))

# Keep words that can reverse or change sentiment
important_words = {
    "no", "nor", "not", "never",
    "don't", "doesn't", "didn't",
    "isn't", "aren't", "wasn't", "weren't",
    "can't", "couldn't", "won't", "wouldn't",
    "shouldn't", "haven't", "hasn't", "hadn't"
}

stop_words = stop_words - important_words

def remove_stopwords(text):
    tokens = word_tokenize(basic_clean(text))

    filtered_tokens = [
        word for word in tokens
        if word.lower() not in stop_words
    ]

    return " ".join(filtered_tokens)

experiment_df['CommentText'].apply(remove_stopwords)
```

```python

import time
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.linear_model import LogisticRegression, SGDClassifier, RidgeClassifier
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB, ComplementNB
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from xgboost import XGBClassifier

# 1. Define the candidate models
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Linear SVM": LinearSVC(),
    "Multinomial Naive Bayes": MultinomialNB(),
    "Complement Naive Bayes": ComplementNB(),
    "SGD Classifier": SGDClassifier(random_state=42),
    "Ridge Classifier": RidgeClassifier()
}

# 2. Use identical CV folds for all models
cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

results = []

# 3. Train and evaluate each model
for name, classifier in models.items():
    print(f"\nTraining: {name}")
    start = time.time()

    pipeline = Pipeline([
        ("tfidf", TfidfVectorizer(
            max_features=100_000,
            ngram_range=(1, 2),
            min_df=2,
            sublinear_tf=True,
            dtype=np.float32
        )),
        ("classifier", classifier)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring={
            "accuracy": "accuracy",
            "macro_f1": "f1_macro"
        },
        n_jobs=1,
        return_train_score=False,
        error_score="raise"
    )

    elapsed = time.time() - start

    results.append({
        "Model": name,
        "CV Accuracy": scores["test_accuracy"].mean(),
        "CV Macro F1": scores["test_macro_f1"].mean(),
        "F1 Std": scores["test_macro_f1"].std(),
        "Time (min)": elapsed / 60
    })

    print(f"Macro F1: {scores['test_macro_f1'].mean():.4f}")
    print(f"Time: {elapsed / 60:.2f} minutes")

# 4. Display the leaderboard
results_df = pd.DataFrame(results)
results_df = results_df.sort_values(
    "CV Macro F1", ascending=False
).reset_index(drop=True)

display(
    results_df.style.format({
        "CV Accuracy": "{:.4f}",
        "CV Macro F1": "{:.4f}",
        "F1 Std": "{:.4f}",
        "Time (min)": "{:.2f}"
    })
)

print("\nBest model:", results_df.iloc[0]["Model"])
print("Best CV Macro F1:", round(results_df.iloc[0]["CV Macro F1"], 4))
```

```python

import time
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV, StratifiedKFold

pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        max_features=100_000,
        ngram_range=(1, 2),
        min_df=2,
        sublinear_tf=True,
        dtype=np.float32
    )),
    ("classifier", LogisticRegression(max_iter=1000))
])

param_grid = {
    "tfidf__max_features": [50_000, 100_000],
    "tfidf__ngram_range": [(1, 1), (1, 2)],
    "classifier__C": [0.5, 1.0, 2.0]
}

cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring="f1_macro",
    cv=cv,
    n_jobs=1,
    verbose=2,
    refit=True
)

start = time.time()
grid_search.fit(X_train, y_train)

print("\nBest parameters:", grid_search.best_params_)
print("Best CV Macro F1:", round(grid_search.best_score_, 4))
print("Time taken (minutes):", round((time.time() - start) / 60, 2))
```

```python
import torch
from transformers import pipeline

MODEL_NAME = "cardiffnlp/twitter-roberta-base-sentiment-latest"

device = 0 if torch.cuda.is_available() else -1

sentiment_model = pipeline(
    task="text-classification",
    model=MODEL_NAME,
    tokenizer=MODEL_NAME,
    device=device
)

print("Model loaded successfully!")
print("Device:", "GPU" if device == 0 else "CPU")
```

```python
comments = [
    "This video is amazing, I really enjoyed it!",
    "The video is okay, nothing special.",
    "This is terrible and completely boring.",
    "I don't like this video at all."
]

predictions = sentiment_model(
    comments,
    batch_size=4,
    truncation=True
)

for comment, result in zip(comments, predictions):
    print("Comment:", comment)
    print("Sentiment:", result["label"])
    print("Confidence:", round(result["score"], 4))
    print("-" * 60)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Select a representative test sample
eval_size = min(5000, len(X_test))

X_hf_eval, _, y_hf_eval, _ = train_test_split(
    X_test,
    y_test,
    train_size=eval_size,
    random_state=42,
    stratify=y_test
)

# Run predictions in batches
hf_results = sentiment_model(
    X_hf_eval.tolist(),
    batch_size=16,
    truncation=True
)

hf_predictions = [r["label"].title() for r in hf_results]

# Normalize actual labels
y_actual = y_hf_eval.astype(str).str.strip().str.title()

print(classification_report(
    y_actual,
    hf_predictions,
    labels=["Negative", "Neutral", "Positive"],
    zero_division=0
))

print("Accuracy:", accuracy_score(y_actual, hf_predictions))
print("Macro F1:", f1_score(
    y_actual, hf_predictions, average="macro", zero_division=0
))
```
